# Cosmological Structure Module Demo
### Gravitational Clustering Visualization

```python
from IPython.display import display
import numpy as np
from PIL import Image
from src.hart_morphosis.core import CosmologicalStructureModule

module = CosmologicalStructureModule(grid_size=(64, 64), G=1.0)
trajectory = module.simulate_clustering(n_clusters=12, steps=25, seed=3)

positions = trajectory[-1].cpu().numpy()
canvas = np.zeros((256, 256, 3), dtype=np.uint8)
for x, y in positions:
    px = int(np.clip(x * 255, 2, 253))
    py = int(np.clip(y * 255, 2, 253))
    canvas[py - 2:py + 3, px - 2:px + 3] = [255, 255, 255]

img = Image.fromarray(canvas)
display(img)
```
