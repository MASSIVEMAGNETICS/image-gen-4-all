# Ruleset Customization
### Editing default.json for new behaviors

```python
import json
from pathlib import Path

rules_path = Path("src/hart_morphosis/models/rulesets/default.json")
rules = json.loads(rules_path.read_text())

rules["morphogenesis"]["feed_rate"] = 0.05
rules["cosmology"]["gravity"] = 1.2

custom_path = Path("src/hart_morphosis/models/rulesets/custom.json")
custom_path.write_text(json.dumps(rules, indent=2))

print(custom_path.read_text())
```
